```python
import os
import cv2
import kagglehub

# Get the correct Kaggle dataset location
dataset_root = kagglehub.dataset_download(
    "belsonraja/face-mask-dataset-with-and-without-mask"
)

# Correct dataset folder
base_path = os.path.join(
    dataset_root,
    "facemask-dataset",
    "dataset"
)

# categories = ["with_mask", "without_mask"]
categories = ["without_mask", "with_mask"]

print("Dataset:", base_path)
print("Categories:", os.listdir(base_path))
```

```python

data=[]
for category in categories:

    folder_path = os.path.join(base_path, category)

    label = categories.index(category)
    
    for file in os.listdir(folder_path):

        img_path = os.path.join(folder_path, file)

        img = cv2.imread(img_path)

        if img is None:
            continue

        img = cv2.resize(img, (224, 224))

        # print(img.shape)
        data.append([img,label])
```

```python

haar = cv2.CascadeClassifier(
    cv2.data.haarcascades + "haarcascade_frontalface_default.xml"
)

print("Loaded:", not haar.empty())
```

```python
# 3. Predict Mask / Without Mask
def detect_face_mask(img):
    img = cv2.resize(img, (224, 224))
    img = img.astype("float32") / 255.0

    y_pred = model.predict(
        img.reshape(1, 224, 224, 3),
        verbose=0
    )

    probability = float(y_pred[0][0])
    prediction = int(probability >= 0.5)

    return prediction, probability
```

```python
def detect_face(gray):
    cords = haar.detectMultiScale(
        gray,
        scaleFactor=1.05,
        minNeighbors=3,
        minSize=(40, 40)
    )
    return cords
```

```python
#5  On video write text
def draw_label(img, text, pos, bg_color):

    text_size = cv2.getTextSize(
        text, cv2.FONT_HERSHEY_SIMPLEX, 1, 2
    )[0]

    end_x = pos[0] + text_size[0] + 2
    end_y = pos[1] + text_size[1] + 2

    cv2.rectangle(
        img, pos, (end_x, end_y),
        bg_color, cv2.FILLED
    )

    cv2.putText(
        img, text, (pos[0], pos[1] + text_size[1]),
        cv2.FONT_HERSHEY_SIMPLEX,
        1, (0, 0, 0), 2, cv2.LINE_AA
    )
```

```python

cap = cv2.VideoCapture(0)

while True:
    ret, frame = cap.read()

    if not ret:
        break

    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    cords = detect_face(gray)
    print("Faces detected:", len(cords))

    if len(cords) > 0:

        # Select the largest detected face
        x, y, w, h = max(
            cords,
            key=lambda face: face[2] * face[3]
        )

        # Crop only the face
        face = frame[y:y+h, x:x+w]

        # Predict on the face, not the entire frame
        prediction, probability = detect_face_mask(face)

        if prediction == 0:
            label = "Without MASK"
            color = (0, 0, 255)
        else:
            label = "MASK"
            color = (0,255, 0 )

        draw_label(frame, label, (10, 10), color)

        cv2.rectangle(
            frame,
            (x, y),
            (x+w, y+h),
            (255, 0, 0),
            2
        )

        print("Prediction:", prediction,
              "Probability:", round(probability, 4))

    else:
        draw_label(
            frame, "FACE NOT DETECTED",
            (10, 10), (0, 255, 255)
        )

    cv2.imshow("window", frame)

    if cv2.waitKey(1) & 0xFF == ord('x'):
        break

cap.release()
cv2.destroyAllWindows()
```
